```python
import pandas as pd
import seaborn as sns
import scipy.stats as stats
from statsmodels.stats import weightstats
```

```python
sns.get_dataset_names()
```

First Dataset

```python
df = sns.load_dataset('taxis')
df = df.sample(frac=0.30)
df.describe()
```

```python
df.head()
```

One-Sample Hypothesis Test for the taxis dataset:

H₀ = The mean tip is equal to $2.\
H₁ = The mean tip is different from $2.

Running the one-sample t-test:

```python
t_stat, p_value = stats.ttest_1samp(df['tip'],2)
print(f"Estatística t: {t_stat}, Valor-p: {p_value}")
```

We fail to reject the null hypothesis. There is not enough evidence to conclude that the mean tip is different from $2.

Two-Sample Hypothesis Test for the taxis dataset:

H₀ = The mean tip paid by credit card is equal to the mean tip paid in cash.\
H₁ = The mean tips paid by credit card and cash are different.

Separating tips by payment method:

```python
tcc = df[df['payment'] == 'credit card']['tip']
tc = df[df['payment'] == 'cash']['tip']
```

Running the two-sample independent t-test:

```python
t_stat_2, p_value_2 = stats.ttest_ind(tcc, tc, equal_var=False)
print(f"Estatística t: {t_stat_2}, Valor-p: {p_value_2}")
```

We reject the null hypothesis. The mean tip paid by credit card is significantly different from the mean tip paid in cash.

Second Dataset

```python
df2 = sns.load_dataset('flights')
df2 = df2.sample(frac=0.30)
df2.describe()
```

```python
df2.head()
```

Removing null values in the 'passengers' column to avoid errors:

```python
df_cleaned = df2.dropna(subset=['passengers', 'year'])
```

One-Sample Hypothesis Test for the flights dataset

H₀ = The mean number of passengers is equal to 300.\
H₁ = The mean number of passengers is different from 300.

Running the one-sample t-test:

```python
t_stat_1, p_value_1 = stats.ttest_1samp(df_cleaned['passengers'], 300)
print(f"Estatística t: {t_stat_1}, Valor-p: {p_value_1}")
```

We fail to reject the null hypothesis. There is not enough evidence to conclude that the mean number of passengers is different from 300.

Two-Sample Hypothesis Test for the flights dataset

H₀ = The mean number of passengers in 1949 is equal to the mean in 1950.\
H₁ = The mean number of passengers in 1949 is different from the mean in 1950.

Filtering the number of passengers for 1949 and 1950:

```python
passengers_1949 = df_cleaned[df_cleaned['year'] == 1949]['passengers']
passengers_1950 = df_cleaned[df_cleaned['year'] == 1950]['passengers']
```

Running the two-sample independent t-test:

```python
t_stat_2, p_value_2 = stats.ttest_ind(passengers_1949, passengers_1950, equal_var=False)
print(f"Estatística t: {t_stat_2}, Valor-p: {p_value_2}")
```

We fail to reject the null hypothesis. There is not enough evidence to conclude that the mean number of passengers between 1949 and 1950 are different.
